## Payment Data Transformation and Reconciliation

### Payment aggregation
- Original payment records: 103,886
- Unique orders with payment records: 99,440
- Aggregated payment records: 99,440
- Zero-value payments: 9
- Negative-value payments: 0
- Orders without payment records: 1

Payments were aggregated by order_id to prevent row multiplication when joining payment information to the order-level dataset.

### Payment and item value reconciliation
- Total item value (product price + freight): 15,843,553.24
- Total recorded payment value: 16,008,872.12
- Orders with matching values (within 0.01): 98,290
- Orders with mismatched values: 1,150

Of the 1,150 mismatched orders:
- 772 had no recorded order-item rows.
- 378 had recorded order items.

Among the 378 orders with item records, the payment-item difference ranged from -51.62 to 182.81.

### Data handling decisions
- Preserved all payment records, including zero-value records.
- Preserved all orders, including orders without payment records.
- Kept payment values and item values as separate business measures.
- Did not force payment values to match item values.
- Did not assume that payments associated with canceled or unavailable orders were refunded or retained.

The payment and item value differences were retained for further business analysis.

```python
# 1. Prepare an order lookup table
order_lookup = orders[
    [
        "order_id",
        "customer_id",
        "order_purchase_timestamp"
    ]
].copy()

order_lookup["purchase_date"] = (
    order_lookup["order_purchase_timestamp"].dt.normalize()
)

order_lookup = order_lookup[
    ["order_id", "customer_id", "purchase_date"]
]


# 2. Create a date-to-key lookup
date_key_lookup = (
    date_dim.set_index("date")["date_key"]
)


# 3. Update FactPayments
fact_payments = fact_payments.merge(
    order_lookup,
    on="order_id",
    how="left",
    validate="many_to_one"
)

fact_payments["purchase_date_key"] = (
    fact_payments["purchase_date"].map(date_key_lookup)
)

fact_payments.drop(columns=["purchase_date"], inplace=True)


# 4. Update FactReviews
fact_reviews = fact_reviews.merge(
    order_lookup,
    on="order_id",
    how="left",
    validate="many_to_one"
)

fact_reviews["purchase_date_key"] = (
    fact_reviews["purchase_date"].map(date_key_lookup)
)

fact_reviews["review_creation_date_key"] = (
    fact_reviews["review_creation_date"]
    .dt.normalize()
    .map(date_key_lookup)
)

fact_reviews["review_answer_date_key"] = (
    fact_reviews["review_answer_timestamp"]
    .dt.normalize()
    .map(date_key_lookup)
)

fact_reviews.drop(columns=["purchase_date"], inplace=True)


# 5. Validate updated fact tables
print("FACT PAYMENTS")
print("Shape:", fact_payments.shape)
print("Missing customer IDs:", fact_payments["customer_id"].isna().sum())
print("Missing purchase date keys:", fact_payments["purchase_date_key"].isna().sum())

print("\nFACT REVIEWS")
print("Shape:", fact_reviews.shape)
print("Missing customer IDs:", fact_reviews["customer_id"].isna().sum())
print("Missing purchase date keys:", fact_reviews["purchase_date_key"].isna().sum())
print("Missing review creation date keys:", fact_reviews["review_creation_date_key"].isna().sum())
print("Missing review answer date keys:", fact_reviews["review_answer_date_key"].isna().sum())
```
