```python
# Find ZIP prefixes associated with multiple states
problem_zip_codes = zip_state_variation[
    zip_state_variation > 1
].index

# Display all records for those ZIP prefixes
geolocation[
    geolocation["geolocation_zip_code_prefix"].isin(problem_zip_codes)
][
    [
        "geolocation_zip_code_prefix",
        "geolocation_lat",
        "geolocation_lng",
        "geolocation_city",
        "geolocation_state"
    ]
].sort_values("geolocation_zip_code_prefix")
```

```python
# Count records belonging to ZIP prefixes with multiple states
problem_records = geolocation[
    geolocation["geolocation_zip_code_prefix"].isin(problem_zip_codes)
]

print("Total records in problematic ZIP prefixes:", len(problem_records))

print(
    "Number of exceptional records:",
    len(problem_records) -
    problem_records.groupby("geolocation_zip_code_prefix").size().sum()
)
```

```python
exceptional_records = []

for zip_code in problem_zip_codes:
    subset = geolocation[
        geolocation["geolocation_zip_code_prefix"] == zip_code
    ]

    dominant_location = subset[
        ["geolocation_city", "geolocation_state"]
    ].value_counts().index[0]

    exceptional = subset[
        (subset["geolocation_city"] != dominant_location[0]) |
        (subset["geolocation_state"] != dominant_location[1])
    ]

    exceptional_records.append(exceptional)

exceptional_geolocation = pd.concat(exceptional_records)

print("Exceptional records:", len(exceptional_geolocation))
```

```python
import pandas as pd
import csv

input_path = r"C:\Users\priya\OneDrive\Desktop\Documents\Downloads\Olist_Data_Warehouse\Dataset\Transformed\FactReviews.csv"

output_path = r"C:\Users\priya\OneDrive\Desktop\Documents\Downloads\Olist_Data_Warehouse\Dataset\Transformed\FactReviews_MySQL.csv"

df = pd.read_csv(input_path)

# Replace line breaks inside review text with spaces
df['review_comment_title'] = (
    df['review_comment_title'].str.replace(r'[\r\n]+', ' ', regex=True)
)

df['review_comment_message'] = (
    df['review_comment_message'].str.replace(r'[\r\n]+', ' ', regex=True)
)

# Save a MySQL-friendly CSV
df.to_csv(
    output_path,
    index=False,
    quoting=csv.QUOTE_ALL,
    doublequote=True,
    lineterminator='\n'
)

print("Original rows:", len(pd.read_csv(input_path)))
print("New CSV rows:", len(pd.read_csv(output_path)))
print("New file created:", output_path)
```
